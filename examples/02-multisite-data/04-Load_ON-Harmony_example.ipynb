# Datalad Example

### This example showcase how you can download and get files from a datalad dataset

```python
from uniharmony import verbosity
from uniharmony.datasets import clean_tmp_folder, download_bids_dataset, load_ONharmony


verbosity("debug")
```

```python
load_ONharmony(
    subjects="16981",
    sessions="NOT4GEP001",
    modalities="anat",
    suffixes="T1w",
    extensions=".json",
    force_download=True,
    hidden=True,
    tmp_clean=False,
)
```

```python
# Later, clean up the temporary cache
clean_tmp_folder("datalad_cache")
```

```python
download_bids_dataset(
    subjects="01",
    sessions="01",
    modalities="eeg",
    tasks="innerspeech",
    suffixes="eeg",
    extensions=".bdf",
    target_path="./Inner",
    dataset_source_URL="https://github.com/OpenNeuroDatasets/ds003626.git",
    runs="all",
    root_files=[],
)
```
